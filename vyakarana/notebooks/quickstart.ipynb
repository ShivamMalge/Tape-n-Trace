```python
import vyakarana
from vyakarana import CFG, DFA, NFA, gallery

vyakarana.__version__
```

```python
d = DFA(
    states={"a", "b"},
    alphabet={"0", "1"},
    transitions={("a", "0"): "b", ("a", "1"): "a", ("b", "0"): "a", ("b", "1"): "b"},
    start="a",
    accepting={"b"},
)
d.accepts("011")  # True: an odd number of 0s
```

```python
run = d.run("011")  # renders the stepped run inline when a frontend is attached
run
```

```python
# The subset construction — the demo §3 of phases-vyakarana.md names.
n = NFA(
    states={"p", "q", "r"},
    alphabet={"0", "1"},
    transitions={("p", "0"): {"p", "q"}, ("p", "1"): {"p"}, ("q", "1"): {"r"}},
    start="p",
    accepting={"r"},
)
sorted(n.to_dfa().states)
```

```python
g = CFG.from_text("S -> a S b | a b")
derivation = g.derive("aabb")
derivation.derived, derivation.steps
```

```python
print(gallery.zeros_ones.run("0011").id_log())  # Fig. 8.9, character for character
```

```python
gallery.never_halts.accepts("1", max_steps=50)  # Halted.NO — a bound, never a silent False
```
