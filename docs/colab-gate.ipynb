```python
from vyakarana import DFA

d = DFA(
    states={"a", "b"},
    alphabet={"0", "1"},
    transitions={("a", "0"): "b", ("a", "1"): "a", ("b", "0"): "a", ("b", "1"): "b"},
    start="a",
    accepting={"b"},
)
d.accepts("011")  # must print True, synchronously — ADR-004's promise
```

```python
d.run("011")  # THE criterion: the run renders inline, with transport controls
```

```python
# Full-surface spot check: the subset construction renders, the TM ID log matches the book.
from vyakarana import NFA, gallery

n = NFA(
    states={"p", "q", "r"},
    alphabet={"0", "1"},
    transitions={("p", "0"): {"p", "q"}, ("p", "1"): {"p"}, ("q", "1"): {"r"}},
    start="p",
    accepting={"r"},
)
n.to_dfa()
print(gallery.zeros_ones.run("0011").id_log())
```
